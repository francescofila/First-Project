Inizializzazione

```python
import torch
from torch import nn

class SimpleNet(nn.Module):
    """Rete neurale fully connected con un solo livello nascosto"""

    def __init__(self):
        """dichisra e registra i layer del modello"""
        super().__init__()

        # Da 10 feature di input a 20 unità nascoste
        self.fc1 = nn.Linear(10, 20)

        # Da 20 unità nascoste a un singolo output
        self.fc2 = nn.Linear(20, 1)


    def forward(self, x):
        """Definiamo il percorso seguito dal tensore in input"""

        # Trasformazione lineare  : (batch, 10) -> (batch, 20)

        x = self.fc1(x)

        # intro non linearità
        x = torch.relu(x)

        # output finale (batch, 20) -> (batch, 1)

        x = self.fc2(x)

        return x
    


model = SimpleNet()
print(model)






```

Creazione di un batch

```python
torch.manual_seed(42)

x = torch.randn(4, 10)
output = model(x)

print("Input:", x.shape)
print("Output:", output.shape)
```

Ispezione  parametri

```python
for name, parameter in model.named_parameters():
    print(
        name,
        "shape =", tuple(parameter.shape),
        "numero =", parameter.numel()
    )
```

```python
total_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

print(total_parameters)
```

Verifica  Autograd

```python
target = torch.randn(4, 1)

criterion = nn.MSELoss()

predictions = model(x)
loss = criterion(predictions, target)

loss.backward()
```

```python
for name, parameter in model.named_parameters():
    print(name, parameter.grad is not None)
```
